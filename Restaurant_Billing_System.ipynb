```python
menu = {
    "Pizza": 200,
    "Burger": 120,
    "Pasta": 150,
    "Sandwich": 80,
    "Coffee": 60,
    "Cold Drink": 50
}

cart = []


def show_menu():

    print("\n========== MENU ==========")

    for item, price in menu.items():
        print(f"{item:15} ₹{price}")

    print("===========================")


def order_food():

    while True:

        show_menu()

        item = input("Enter item name (or 'done'): ")

        if item.lower() == "done":
            break

        if item not in menu:
            print("Item not available.")
            continue

        quantity = int(input("Enter quantity: "))

        amount = menu[item] * quantity

        cart.append({
            "item": item,
            "quantity": quantity,
            "amount": amount
        })

        print("Item added!")


def generate_bill():

    if not cart:
        print("No items ordered.")
        return

    print("\n========== BILL ==========")

    subtotal = 0

    for item in cart:

        print(
            f'{item["item"]} x {item["quantity"]} = ₹{item["amount"]}'
        )

        subtotal += item["amount"]

    gst = subtotal * 0.05
    total = subtotal + gst

    print("--------------------------")
    print("Subtotal : ₹", subtotal)
    print("GST 5%   : ₹", gst)
    print("Total    : ₹", total)
    print("==========================")


def main():

    while True:

        print("""
===== RESTAURANT BILLING =====

1. Order Food
2. Generate Bill
3. Exit
""")

        choice = input("Enter choice: ")

        if choice == "1":
            order_food()

        elif choice == "2":
            generate_bill()

        elif choice == "3":
            break

        else:
            print("Invalid choice!")


main()
```
