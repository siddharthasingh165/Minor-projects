```python
import random

while True:
    choice = input("Choose Heads or Tails: ").lower()

    result = random.choice(["heads", "tails"])

    print("Coin:", result)

    if choice == result:
        print("You Win! 🎉")
    else:
        print("You Lose!")

    again = input("Play again? (y/n): ")

    if again.lower() != "y":
        break
```
