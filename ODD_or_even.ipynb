```python
import random

score = 0

for i in range(5):
    number = random.randint(1, 100)
    choice = input("Even or Odd? ").lower()

    print("Number:", number)

    if (number % 2 == 0 and choice == "even") or \
       (number % 2 != 0 and choice == "odd"):
        print("Correct!")
        score += 1
    else:
        print("Wrong!")

print("Final Score:", score, "/ 5")
```
