```python
import urllib.request
import json

url = "https://random-word-api.herokuapp.com/word?number=1&length=5"

try:
    response = urllib.request.urlopen(url)
    data = json.loads(response.read())

    word = data[0].lower()

    hidden = ["_"] * len(word)

    # Show first two letters
    hidden[0] = word[0]
    hidden[1] = word[1]

    attempts = 6
    guessed = []

    print("WORD GUESSING GAME")
    print("------------------")
    print("Two letters are already revealed!")

    while attempts > 0:

        print("\nWord:", " ".join(hidden))
        print("Guessed letters:", " ".join(guessed))
        print("Attempts left:", attempts)

        letter = input("Enter a letter: ").lower().strip()

        if len(letter) != 1 or not letter.isalpha():
            print("Please enter only one letter.")
            continue

        if letter in guessed:
            print("You already guessed this letter!")
            continue

        guessed.append(letter)

        if letter in word:
            print("Correct!")

            for i in range(len(word)):
                if word[i] == letter:
                    hidden[i] = letter

        else:
            print("Wrong!")
            attempts -= 1

        if "_" not in hidden:
            print("\nYou Won!")
            print("The word was:", word)
            print("Attempts remaining:", attempts)
            break

    else:
        print("\nGame Over!")
        print("The word was:", word)

except Exception as e:
    print("Error:", e)
```
